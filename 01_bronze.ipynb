```python
dbutils.widgets.text("catalog", "scl")
catalog = dbutils.widgets.get("catalog")
```

#customers

```python
spark.sql(f"""
CREATE OR REPLACE TABLE {catalog}.bronze.customer
COMMENT "Customer table"
AS
SELECT *,
    _metadata.file_path as source_path,
    current_timestamp() as ingested_at
FROM samples.tpch.customer
""")
```

```python
spark.sql(f"SELECT * FROM {catalog}.bronze.customer").count()
```

```python
display(spark.sql(f"SELECT * FROM {catalog}.bronze.customer LIMIT 5"))
```
